```python
# counting vowels in a multi-line string
text="""This is a multi-line string.
It can span multiple lines.
It is often used for documentation or long text."""
count=0
for vowel in text:
    if vowel in 'aeiouAEIOU':
        count += 1
print("Number of vowels in the text:", count)
```

```python
# guessing game
import random
# generating 5 random numbers between 1 and 100
secret_number = random.randint(1,100)
while True:
    guess = int(input("Guess the secret number between 1 and 100: "))
    if guess == secret_number:
        print("Congratulations! You guessed the secret number.")
        break
    elif guess < secret_number:
        print("Too low! Try again.")
    elif guess > secret_number:
        print("Too high! Try again.")
    else:
        print("Congratulations! You guessed the secret number.")
        break
```

```python
# password strength checker
has_upper=False
has_lower=False
has_digit=False
password=input("Enter a password: ")
for char in password:
    if char.isupper():
        has_upper=True
    elif char.islower():
        has_lower=True
    elif char.isdigit():
        has_digit=True
if has_upper and has_lower and has_digit:
    print("Password is strong.")
else:
    print("Password is weak. It should contain uppercase, lowercase letters and digits.")
```

```python
text = input("Enter sentence: ")
words = text.split()
print(words)
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

for word, count in freq.items():
    print(word, ":", count)
```

```python
from itertools import count
for i in count(1):
    print(i)
    if i >= 20:
        break
```

```python
from itertools import combinations
items = ['A', 'B', 'C']
for combo in combinations(items, 2):
    print(combo)
```

```python
from itertools import repeat
for item in repeat("Hello", 5):
    print(item)
```

```python
from itertools import permutations
items = ['A', 'B', 'C']
for perm in permutations(items):
    print(perm)
```

```python
from itertools import chain
list1 = [1, 2, 3]
list2 = ['A', 'B', 'C']
for item in chain(list1, list2):
    print(item)
```

```python
from itertools import product
list1 = [1, 2]
list2 = ['A', 'B']
for pair in product(list1, list2):
    print(pair)
```
